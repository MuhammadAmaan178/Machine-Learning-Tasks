```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="Target")
```

```python
X.info()
```

```python
X.describe()
```

```python
y.info()
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
X_train.shape, X_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier()
```

```python
rf.fit(X_train,y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
rf_accuracy = accuracy_score(y_test,y_pred)
rf_accuracy
```

```python
rf_ps = precision_score(y_test,y_pred)
rf_ps
```

```python
rf_rs = recall_score(y_test,y_pred)
rf_rs
```

```python
rf_f1 = f1_score(y_test,y_pred)
rf_f1
```

```python
confusion_matrix(y_test,y_pred)
```

```python
sns.heatmap(confusion_matrix(y_test,y_pred),annot=True, fmt=".0f")
plt.title("Random Forest - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dc = DecisionTreeClassifier()
```

```python
dc.fit(X_train,y_train)
```

```python
dc_pred = dc.predict(X_test)
```

```python
dc_accuracy = accuracy_score(y_test,dc_pred)
dc_accuracy
```

```python
dc_ps = precision_score(y_test,dc_pred)
dc_ps
```

```python
dc_rs = recall_score(y_test,dc_pred)
dc_rs
```

```python
dc_f1 = f1_score(y_test,dc_pred)
dc_f1
```

```python
confusion_matrix(y_test,dc_pred)
```

```python
sns.heatmap(confusion_matrix(y_test,dc_pred),annot=True, fmt=".0f")
plt.title("Decision Tree - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Model": ["Random Forest", "Decision Tree"],
    "Accuracy": [rf_accuracy, dc_accuracy],
    "Precision": [rf_ps, dc_ps],
    "Recall": [rf_rs, dc_rs],
    "F1-Score": [rf_f1, dc_f1]
})
```
